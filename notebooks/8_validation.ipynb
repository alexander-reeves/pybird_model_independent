# Validation: extended cosmologies, θ_s, noiseless mocks, neutrino mass
Template `bao_template_no_rs` (the baseline). **Direct fit**: the full likelihood sampled in the cosmological parameters (`run.py reference`). **MI**: the model-independent posterior projected onto the same model with the same priors (`run.py project`, score regression). A reconstruction that loses nothing gives shift 0 and width ratio 1; shifts are in direct-fit σ. DESI-alone priors: BBN ω_b, n_s ~ N(0.9649, 0.042).

θ_s: 100θ_s ~ N(1.040268, 0.000246). Where no direct chain was sampled with it, the DESI-alone direct draws are weighted by that prior (effective number printed). EDE and neutrino mass: θ_s = the ΛCDM value × the ede-v2 emulator's ratio to its ΛCDM limit (`cosmology.theta_s`); for EDE the prior is a test of the projection, not a CMB constraint.

w0waCDM is compared with the direct fit that has the EFT priors in sound-horizon units, which is what α_rs = 1 means (7_alpha_rs_comparison.ipynb); the h/Mpc-prior fit differs from it by 0.5σ.

## 2. Noiseless mocks with a known truth
Data vector = the true likelihood's prediction at the truth (`config.MOCKS`: the cosmology and EFT parameters of the data's ΛCDM best fit with the extension switched on), DR1 covariance, priors and sampler unchanged; the θ_s prior is centred on the truth's θ_s. Dotted lines: the truth. A noiseless mock tests bias, not coverage: (direct − truth)/σ is the prior-volume and prior-centre effect of the standard analysis itself, MI − direct is what the reconstruction adds; if the projection reproduces the direct posterior, its coverage is that of the direct analysis.
### w0waCDM truth: w0 = −0.75, wa = −0.9

```python
P.use('mock_w0wa'); P.chain_table()
```

The MI posterior against the MI parameters of the truth. The α are unbiased. a(k) and f sit 1.4–2σ from the truth in opposite directions: b1² a and f/b1 are measured, so a, b1 and f move together, and the marginals follow the prior volume along that direction. The measured combination f D a^1/2 is 3–5 times tighter and within 0.5–0.8σ, the size of the direct fit's own offset in ln A_s below. Read the a(k) and f marginals of the reconstruction with this in mind; the projections do not use them.

```python
P.mi_truth();
```

## 3. Neutrino mass
`mnu`: ΛCDM + m_ν, the mass of each of three degenerate neutrinos (Σm_ν = 3 m_ν), flat in [0, 0.33] eV. Direct fit: the cdm + baryon spectrum of the emulator at each redshift, so with the scale-dependent growth. MI: one spectrum at z = 1.49 times scale-independent growth ratios, f and the distances with the neutrinos counted as matter. A bias of the reconstruction shows as MI − direct.
