## Why α_rs = 1 moves w0waCDM: the units of the EFT priors
$T_{1/\alpha_{r_s}}$ (notebook 6) takes the point of a cosmology in `bao_template` to $\alpha_{r_s}=1$: $\alpha\to\alpha/\alpha_{r_s}$, $P_{\rm lin}(k)\to\alpha_{r_s}^{-3}P_{\rm lin}(k/\alpha_{r_s})$, EFT $c\to\alpha_{r_s}^{-n}c$ ($n=0$ for $b_3$; 2 for $c_{ct},c_{r1},c_{r2}$; 3 for $c_{\epsilon0}$; 5 for $c_{\epsilon1},c_{\epsilon2}$), $\alpha_{r_s}=h r_d/(h r_d)_{\rm fid}$ of the cosmology. The data term does not change; the Gaussian priors $N(\mu,\sigma)$ of the 42 profiled $c$ stay where they are:

$$\Delta\chi^2=\sum\frac{(\alpha_{r_s}^{-n}c-\mu)^2-(c-\mu)^2}{\sigma^2}\simeq-2\ln\alpha_{r_s}\sum\frac{n\,c\,(c-\mu)}{\sigma^2}.$$

So `bao_template_no_rs` is the same likelihood with the EFT priors in units of $r_d$ instead of $h^{-1}$Mpc.

Test (`checks.py exact_repr`): 2000 draws of the w0waCDM direct chain, importance-weighted to each likelihood. `oracle` = the exact spectrum of the draw in the $\alpha_{r_s}=1$ frame, no template. If the priors are the cause: oracle − truth equals the formula draw by draw (evaluated at the truth's profiled $c$; re-profiling can only lower it), the same with the loop emulator and with exact loops; it vanishes with the priors × 1000; and `bao_template_no_rs` − oracle ≈ 0.

The converse (`checks.py adapted`): for each draw take its own $\alpha_{r_s}$ and give the $\alpha_{r_s}=1$ frame the h/Mpc priors, $N(\alpha_{r_s}^{-n}\mu,\alpha_{r_s}^{-n}\sigma)$ on the frame's coefficients. Rows `_adapted`: the shift against the normal-prior likelihood drops to the 0.1σ that α_rs free and the priors × 1000 also leave.

### The direct fit with the EFT priors in sound-horizon units
`run.py reference --model w0wa --sound-horizon` (`templates.sound_horizon_loglkl`): the true likelihood of the cosmology, no template, written in the α_rs = 1 frame. pybird gets the cosmology's own spectrum with every length in units of its sound horizon, $\tilde P(k)=\rho^3P(\rho k)$ with $\rho=r_{d,\rm fid}/r_d$ (the BAO wiggles at their fiducial positions), and the BAO α, $(D_H/r_d)/(D_H/r_d)_{\rm fid}$ and $(D_M/r_d)/(D_M/r_d)_{\rm fid}$, as its AP dilations; f and D as before. The likelihood object, the EFT parameters and their priors are not touched. That prediction equals the standard one with $c\to\alpha_{r_s}^{-n}c$, so the unchanged Gaussian priors now sit on EFT coefficients whose scales ($k_M$, $k_R$, $\bar n$) are in units of $r_d$.

All together: both direct fits and both templates projected. α_rs free follows the h/Mpc fit, α_rs = 1 the sound-horizon fit.
